# Using previous results in calculations

`wc` and `wnlc` save the last successful calculation for use in the next calculation.

When using `wc`, Wolfram stores its input as `lhs` (left hand side) and its evaluated output as `rhs` (right hand side). The `lhs` will be held from evaluation such that it does not merely simplify to the right hand side result stored in `rhs`.

`wnlc` allows referencing the result (right hand side) of the last expression as `!result` such that operations specified in natural language can be applied to the result of the last computation.

If it is desired to apply a new computation specified in natural language to both sides (left and right hand side) of the previous computation, the special `!bothsides` directive can be used in `wnlc` calls, e.g., *"Integrate !bothsides"*.

```python
from pyderivehelper import wc, wnlc
```

## Use `rhs` in `wc`

The second call substitutes into the result of the first call. Here `rhs` is the expanded polynomial.

```python
wc('Expand[(x + 1)^2]')
wc('rhs /. x -> 2')
```

## Use held `lhs` in `wc`

`lhs` preserves the previous Wolfram input. Use `ReleaseHold[lhs]` when you want to evaluate it in a new calculation. The first call below resets the example so `lhs` refers to `Expand[(x + 1)^2]`.

```python
wc('Expand[(x + 1)^2]')
wc('ReleaseHold[lhs] /. x -> 2')
```

## Use `!result` in `wnlc`

In a natural-language prompt, `!result` refers to the previous `rhs`. The prompt asks for an operation on that result alone.

```python
wc('Sin[x]')
wnlc('Differentiate !result with respect to x')
```

## Use `!bothsides` in `wnlc`

`!bothsides` applies the requested operation to the previous `lhs` and `rhs` separately, then displays their results as an equation. The first call establishes the equation `Expand[(x + 1)^2] = x^2 + 2 x + 1`.

```python
wc('Expand[(x + 1)^2]')
wnlc('Integrate !bothsides with respect to x')
```

## Continue with the new result

The two-sided calculation becomes the latest reference point. Its displayed left side is held in `lhs`, and its right side is in `rhs`, so `wc` can use the result of `wnlc` immediately.

```python
wc('rhs /. x -> 0')
```

## Example derivation

Using `!result` and `!bothsides` allows carrying out successive steps of a derivation in natural language. Combining with `/calc` allows quickly outputting multiple calculation steps.

```python
wnlc('/calc Integrate sin(x) with respect to x')
wnlc('/calc Integrate the !result with respect to x')
wnlc('/calc Add x^2 to the !result')
```

`/calc` can be combined with `/showtex` and `/showcode` to show `TeX` and Wolfram code for select steps.

```python
wnlc('/calc /showtex Integrate sin(x) with respect to x')
wnlc('/calc /showtex Integrate the !result with respect to x')
wnlc('/calc /showtex /showcode Add x^2 to the !result')
```
