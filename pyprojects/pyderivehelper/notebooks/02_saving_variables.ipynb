# Saving named results

Use `/saveas !name` in an evaluating `wnlc` prompt to save its result. Later prompts can refer to that result as `!name`, even after `!result` has changed. A prompt can contain one `/saveas` command with multiple names separated by spaces or tabs. Names may contain letters, digits, and underscores, but must start with a letter or underscore.

```python
from pyderivehelper import wnlc, ws
```

## Save and reuse one result

The first calculation saves its result as `!sinIntegral`. The next calculation changes the latest `!result`, but the saved name still refers to the integral.

```python
wnlc('/calc Integrate sin(x) with respect to x /saveas !mySinIntegral')
```

```python
wnlc('/calc Expand (x + 1)^2')
wnlc('/calc Differentiate !mySinIntegral with respect to x')
```

## Save one result under multiple names

Both names below refer to the same saved polynomial. The `/saveas` command is removed before the calculation is sent to the model.

```python
wnlc('/calc Expand (x + 1)^2 /saveas !poly1 !expandedPoly')
```

```python
from pyderivehelper.wltools import set_log_level

set_log_level('DEBUG')

wnlc('/calc Evaluate !poly1 at x = 2')
wnlc('/calc Differentiate !expandedPoly with respect to x')
```
